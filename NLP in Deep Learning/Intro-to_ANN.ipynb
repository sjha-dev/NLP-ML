# Artificial Neural Network (ANN)

## What is ANN?

An **Artificial Neural Network (ANN)** is a Machine Learning model inspired by the structure of the **human brain**. It learns patterns from data by adjusting **weights** and **biases** during training.

---

# Applications of ANN

### 1. Classification
Used when the output belongs to a fixed category.

**Examples:**
- Spam Detection
- Sentiment Analysis
- Text Classification
- Disease Prediction
- Cat vs Dog Classification

**Output Example:**
```
Input : "This movie is amazing!"
Output: Positive 😊
```

---

### 2. Regression
Used when the output is a continuous numerical value.

**Examples:**
- House Price Prediction
- Salary Prediction
- Temperature Forecasting
- Stock Price Prediction

**Output Example:**
```
Input : House Features
Output: ₹52,50,000
```

---

# Example: House Price Prediction

### Input Features (X)

| Feature | Description |
|----------|-------------|
| f1 | House Size |
| f2 | Number of Rooms |

### Output (y)

| Target |
|---------|
| House Price |

Example Dataset

| House Size | Rooms | Price |
|------------|-------|--------|
| 1000 sqft | 2 | ₹40L |
| 1500 sqft | 3 | ₹65L |
| 2000 sqft | 4 | ₹90L |

---

# Features

Features are the **input variables** given to the model.

Examples:
- House Size
- Number of Rooms
- Age of House
- Parking Space

---

# Target (Label)

The value that the model has to predict.

```
Target = y
```

Example:
```
House Price
```

---

# Tabular Data

Data stored in **rows and columns** is called **Tabular Data**.

Example:

| Size | Rooms | Price |
|------|------|------|
|1200|2|45L|
|1500|3|70L|

ANN works well on tabular datasets.

---

# Structure of ANN

```
Input Layer
      │
      ▼
Hidden Layer(s)
      │
      ▼
Output Layer
```

### Input Layer
Receives the input features.

Example:
- House Size
- Number of Rooms

---

### Hidden Layer
Learns patterns and relationships between features automatically.

Example:
```
Large House
+
More Rooms
↓

Higher Price
```

---

### Output Layer

Produces the final prediction.

Examples:

Regression
```
₹55 Lakh
```

Classification
```
Positive
Spam
Cat
```

---

# Weights (W)

Each connection between neurons has a **Weight**.

Weights determine the importance of each feature.

Example:

```
House Size → Weight = 0.8
Rooms      → Weight = 0.2
```

This means **House Size** has more influence on the prediction.

---

# Bias (b)

Bias is an additional parameter added to every neuron.

Formula:

```
Output = Σ(Weight × Input) + Bias
```

Bias makes the model more flexible and improves learning.

---

# Forward Propagation

Forward Propagation is the process of passing inputs through the network to generate a prediction.

```
Input
   ↓
Hidden Layer
   ↓
Output (ŷ)
```

---

# Predicted Output (ŷ)

The output predicted by the model.

Example

```
Actual Price     = ₹50 Lakh
Predicted Price  = ₹48 Lakh
```

Here,

```
ŷ = ₹48 Lakh
```

---

# Actual Output (y)

The correct value present in the dataset.

Example

```
y = ₹50 Lakh
```

---

# Loss Function

Loss measures the difference between the **Actual Output (y)** and the **Predicted Output (ŷ)**.

```
Loss = Difference(y, ŷ)
```

Example

```
Actual Price     = ₹50 Lakh
Predicted Price  = ₹48 Lakh

Loss = ₹2 Lakh
```

The objective is to **minimize the loss**.

---

# Backpropagation

Backpropagation is the learning algorithm used to update the model parameters.

Steps:

```
Prediction
      ↓
Calculate Loss
      ↓
Compute Error
      ↓
Update Weights & Biases
      ↓
Better Prediction
```

This process repeats until the loss becomes very small.

---

# Goal of Training

```
Iteration 1 → Loss = 25

Iteration 2 → Loss = 12

Iteration 3 → Loss = 6

Iteration 4 → Loss = 2

Iteration 5 → Loss = 0.5
```

Goal:

```
Loss ↓
Accuracy ↑
```

---

# Sequence of Data Doesn't Matter

ANN assumes that every training example is **independent**.

Example:

```
1000 sqft
1500 sqft
2000 sqft
```

or

```
2000 sqft
1000 sqft
1500 sqft
```

Both are acceptable because the order of rows does not affect learning.

This is why ANN is suitable for **Tabular Data** but **not ideal for sequential data like text or time series**, where order matters.

---

# ANN Workflow

```
Input Features (X)
        │
        ▼
Input Layer
        │
        ▼
Hidden Layer(s)
        │
        ▼
Output Layer
        │
        ▼
Predicted Output (ŷ)
        │
        ▼
Loss Calculation
        │
        ▼
Backpropagation
        │
        ▼
Update Weights & Biases
        │
        ▼
Repeat until Loss is Minimum
```

---

# Quick Revision

- ANN is inspired by the human brain.
- Used for **Classification** and **Regression** tasks.
- Best suited for **Tabular Data**.
- Inputs are called **Features (X)**.
- Output is called **Target (y)**.
- Each connection has a **Weight (W)**.
- Every neuron has a **Bias (b)**.
- **Forward Propagation** generates predictions.
- **Loss Function** measures prediction error.
- **Backpropagation** updates weights and biases.
- Training continues until the **loss is minimized**.
- ANN assumes that the **sequence of data does not matter**.

# CNN (Convolutional Neural Network)

## What is CNN?

A **Convolutional Neural Network (CNN)** is a Deep Learning model specifically designed to process **Images, Videos, and Visual Data**.

Unlike ANN, CNN can automatically learn important visual features such as **edges, textures, shapes, and objects** from an image.

---

# Applications of CNN

- Image Classification
- Object Detection
- Face Recognition
- Medical Image Analysis
- OCR (Optical Character Recognition)
- Self-Driving Cars
- Video Analysis
- Image Segmentation

---

# Image Classification

Image Classification predicts **what object is present** in an image.

### Example

```
Input Image
      ↓
Model
      ↓
Output: Cat
```

Another Example

```
Image
   ↓
Dog
Car
Flower
Human
```

---

# Object Detection

Object Detection predicts:

- What object is present?
- Where is the object located?

### Example

```
Input Image
        ↓
Person
Car
Dog

+ Bounding Boxes around each object
```

---

# CNN vs ANN

| ANN | CNN |
|------|------|
| Works on Tabular Data | Works on Images & Videos |
| Input is Numerical Features | Input is Pixel Values |
| Cannot capture spatial information efficiently | Captures spatial patterns using Convolution |
| Example: House Price Prediction | Example: Face Recognition |

---

# Sequential Data

## What is Sequential Data?

Sequential Data is data in which **the order of elements is important**.

Changing the order changes the meaning.

### Example

```
The food is good
```

If we change the order

```
Good is food the
```

The meaning changes completely.

---

# Why Sequence Matters?

Example

```
I love AI
```

If reordered

```
AI love I
```

The sentence becomes meaningless.

Therefore,

```
Order Matters ✅
```

---

# Applications of Sequential Data

- Text Generation
- Chatbots
- Machine Translation
- Speech Recognition
- Auto Suggestion
- Language Modeling
- Time Series Forecasting

---

# Time Series Data

Time Series Data is data collected over time.

Here, **time order is very important**.

Examples

- Sales Data
- Stock Prices
- Weather Data
- Electricity Consumption
- Website Traffic
- Temperature Records

---

# Sales Forecasting

Sales Forecasting predicts future sales using historical sales data.

Example Dataset

| Date | Sales |
|------|------:|
| 1 Jan | 100 |
| 2 Jan | 120 |
| 3 Jan | 150 |
| 4 Jan | 170 |

Prediction

```
Past Sales
100
120
150
170

↓

Future Sales

190
200
```

The model learns the trend from previous dates to estimate future values.

---

# Why Time Order Matters?

Correct Order

```
1 Jan
2 Jan
3 Jan
4 Jan
```

Incorrect Order

```
4 Jan
2 Jan
1 Jan
3 Jan
```

The second sequence breaks the trend and leads to poor predictions.

# Quick Revision

- **CNN** is designed for **Images and Videos**.
- CNN is widely used for **Image Classification** and **Object Detection**.
- **Sequential Data** is data where the **order of elements matters**.
- Examples of sequential data include **text, speech, chatbot conversations, and time-series data**.
- **Text Generation** predicts the next word in a sentence.
- **Chatbots** generate answers by understanding the sequence of previous words.
- **Auto Suggestion** predicts the next word while typing.
- **Time Series Data** contains a time component, making the order of observations important.
- **Sales Forecasting** predicts future sales using historical sales trends.
- Choose the model based on the type of data:
  - **ANN → Tabular Data**
  - **CNN → Images & Videos**
  - **RNN/LSTM/GRU/Transformer → Sequential Data & Time Series**
