# RNN (Recurrent Neural Network)

**RNN (Recurrent Neural Network)** is a type of **Deep Learning model** mainly used for **sequential data**.

Sequential data is data where the **order of information matters**.

### Examples
- 📈 **Time Series** → Stock prices, temperature, sales
- 📝 **Text** → Words in a sentence
- 🎙️ **Speech** → Audio sequences
- 📡 **Sensor Data** → Heart rate, IoT sensor readings

---

## 🧠 Basic Idea of RNN

Unlike a normal Neural Network, an RNN can use information from **previous time steps** while processing the current input.

```text
       x₁           x₂           x₃           x₄
       ↓            ↓            ↓            ↓
    ┌──────┐     ┌──────┐     ┌──────┐     ┌──────┐
    │ RNN  │ ──→ │ RNN  │ ──→ │ RNN  │ ──→ │ RNN  │
    │ Cell │     │ Cell │     │ Cell │     │ Cell │
    └──┬───┘     └──┬───┘     └──┬───┘     └──┬───┘
       ↓            ↓            ↓            ↓
      h₁           h₂           h₃           h₄

              Previous information
                  flows forward

 ```
## Each RNN Cell    

                  Current Input
                  xₜ
                   ↓
                   +
                   ↓
Previous ─────→ [ RNN ] ─────→ Output
Hidden State       │             yₜ
   hₜ₋₁            ↓
              New Hidden State
                   hₜ






## The basic mathematical representation is:

        hₜ = f(Wₓxₜ + Wₕhₜ₋₁ + b)

where :

xₜ → Current input
hₜ₋₁ → Previous hidden state
hₜ → Current hidden state
Wₓ → Weight for current input
Wₕ → Weight for previous hidden state
b → Bias
f → Activation function


This allows the RNN to remember information from earlier time steps.
