### Word2Vec is a neural network-based embedding technique that converts words into dense vector representations. It learns semantic relationships by predicting target words from their surrounding context (CBOW) or predicting surrounding words from a target word (Skip-Gram). Words appearing in similar contexts obtain similar vector representations, making Word2Vec highly effective for similarity search, text classification, clustering, and many other NLP tasks.

## Key Features

- Converts words into dense numerical vectors.
- Captures semantic and syntactic relationships.
- Similar words have similar embeddings.
- More efficient and meaningful than Bag of Words and TF-IDF.

## Word2Vec Architectures

### 1. Continuous Bag of Words (CBOW)
- Predicts the **target word** using its surrounding **context words**.
- Faster to train and works well on large datasets.

### 2. Skip-Gram
- Predicts the **context words** from a given **target word**.
- Performs better for small datasets and rare words.

## Advantages

- Captures semantic similarity between words.
- Produces low-dimensional dense vectors.
- Improves the performance of many NLP tasks.
- Efficient to train on large corpora.

## Limitations

- Generates only one embedding per word (context-independent).
- Cannot represent multiple meanings of the same word.
- Requires retraining for unseen words.

## Applications

- Text Classification
- Sentiment Analysis
- Machine Translation
- Information Retrieval
- Chatbots
- Recommendation Systems

```python
## implementation of word2vec

from gensim.models import Word2Vec
```

```python
corpus = [
    "students study data structures and algorithms",
    "leetcode helps improve problem solving skills",
    "consistent practice leads to better coding skills",
    "internships provide real world experience",
    "software engineers solve complex problems"
]


tokens = [sentence.split() for sentence in corpus]
model=  Word2Vec(
    sentences=tokens,
    vector_size=100,
    window=2,
    min_count=1
)


```

```python
print (model.wv["real"])
```

```python
print(model.wv.most_similar("leads"))
```
