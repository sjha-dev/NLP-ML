# N-Grams

## What are N-Grams?

An **N-Gram** is a sequence of **N consecutive words** from a sentence. It helps preserve the **word order** and **local context**, unlike the Bag of Words (BoW) model.

---

## Example

Sentence:

```text
I love machine learning
```

| N-Gram Type | Output |
|-------------|--------|
| Unigram (N=1) | I, love, machine, learning |
| Bigram (N=2) | I love, love machine, machine learning |
| Trigram (N=3) | I love machine, love machine learning |

---

## Advantages over Bag of Words (BoW)

- Preserves the **order of words**.
- Captures **local context** between neighboring words.
- Recognizes meaningful phrases such as **"New York"** and **"not bad"**.
- Often improves performance in tasks like **sentiment analysis** and **text classification**.

---

## Limitations

- The number of features increases rapidly as **N** grows (**feature explosion**).
- Requires more memory and computational resources.
- Produces sparse feature vectors for large datasets.

---


# Applications of N-Grams

- Sentiment Analysis
- Text Classification
- Spam Detection
- Machine Translation
- Search Engines
- Auto-Completion
- Next Word Prediction
- Language Modeling

---


## Summary

- **Unigram (N=1):** Single word
- **Bigram (N=2):** Two consecutive words
- **Trigram (N=3):** Three consecutive words

N-Grams are an improvement over Bag of Words because they retain **word order** and **context**, but they become computationally expensive for large vocabularies.

````python
from sklearn.feature_extraction.text import CountVectorizer
````

````python
corpus = [
    "The food is good ",
    "The food is not good "
]
````

````python
cv = CountVectorizer(ngram_range= (2,2))       ## only bigram 
x= cv.fit_transform(corpus)
print(cv.get_feature_names_out())
print(x.toarray())
````

````python
## here the ngram preserve the local context and words ordering , but there are increase in vocabulary and features in ngram.
````

````python
cv = CountVectorizer(ngram_range=(1,2))       ## unigram + bigram 
x= cv.fit_transform(corpus)
print(cv.get_feature_names_out())
print(x.toarray())
````
