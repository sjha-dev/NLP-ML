# Word Embeddings

## Definition
Word Embeddings are **dense vector representations** of words where semantically similar words have similar vector values. Unlike **One-Hot Encoding** and **Bag of Words**, they capture the **meaning and context** of words.

---

## Why Word Embeddings?
Traditional text representations have some limitations:
- High-dimensional sparse vectors
- No semantic relationship between words
- Cannot understand context

Word Embeddings solve these problems by learning meaningful numerical representations.

---

## Example

| Word | Embedding (5-D Example) |
|------|--------------------------|
| King | [0.82, 0.15, -0.44, 0.73, 0.11] |
| Queen | [0.79, 0.18, -0.41, 0.76, 0.14] |
| Man | [0.65, 0.10, -0.30, 0.55, 0.08] |
| Woman | [0.63, 0.13, -0.27, 0.58, 0.12] |

Here, **King** and **Queen** have similar embeddings because their meanings are related.

---

## Advantages
- Captures semantic meaning
- Dense and low-dimensional vectors
- Similar words have similar representations
- Improves NLP model performance

---

## Limitations
- Static embeddings (Word2Vec, GloVe) cannot handle multiple meanings of a word.
- Require large datasets for training.

---

## Popular Word Embedding Techniques
- **Word2Vec**
- **GloVe**
- **FastText**
- **BERT (Contextual Embeddings)**

---

## Applications
- Sentiment Analysis
- Machine Translation
- Text Classification
- Chatbots
- Search Engines
- Question Answering

---

## Key Takeaway
**Word Embeddings convert words into dense numerical vectors that preserve semantic meaning, allowing NLP models to understand relationships between words effectively.**
