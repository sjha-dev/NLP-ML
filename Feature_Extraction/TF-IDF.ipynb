# TF-IDF (Term Frequency - Inverse Document Frequency)

## What is TF-IDF?

**TF-IDF** is a statistical technique used to measure how important a word is in a document relative to an entire collection of documents (corpus).

Unlike **Bag of Words (BoW)**, TF-IDF reduces the importance of very common words and gives higher importance to rare but meaningful words.

---

## Formula

### 1. Term Frequency (TF)

Measures how frequently a term appears in a document.

\[
TF = \frac{\text{Number of times a word appears in a document}}{\text{Total number of words in the document}}
\]

### 2. Inverse Document Frequency (IDF)

Measures how unique a word is across all documents.

\[
IDF = \log\left(\frac{\text{Total number of documents}}{\text{Number of documents containing the word}}\right)
\]

### 3. TF-IDF

\[
TF\text{-}IDF = TF \times IDF
\]

---

## Example

Documents:

```text
D1: I love machine learning
D2: I love deep learning
D3: I love programming
```

- **"love"** appears in every document → Low TF-IDF score.
- **"machine"** appears only in D1 → High TF-IDF score.
- **"deep"** appears only in D2 → High TF-IDF score.

Thus, TF-IDF assigns **higher weights to unique words** and **lower weights to common words**.

---

## Advantages over Bag of Words

- Reduces the importance of frequently occurring words.
- Gives higher importance to informative and unique words.
- Produces better feature representation for text classification.
- Improves model performance compared to simple word counts.

---

## Limitations

- Ignores word order and context.
- Cannot capture semantic meaning.
- Creates sparse feature vectors.
- Does not understand synonyms (e.g., "car" and "automobile").

---

## Summary

- **TF** measures how often a word appears in a document.
- **IDF** measures how rare a word is across all documents.
- **TF-IDF = TF × IDF**
- Words common across many documents receive **low scores**.
- Rare but informative words receive **high scores**.

````python
from sklearn.feature_extraction.text import TfidfVectorizer
````

````python
corpus = [
    "The cat sat on the mat",
    "The dog sat on the rug",
    "The cat chased the mouse"
]
````

````python
vectorizer = TfidfVectorizer()
x = vectorizer.fit_transform(corpus)
print("features")
print(vectorizer.get_feature_names_out())
print(x.toarray())
````
