#  Bag of Words (BoW)

## Definition

Bag of Words (BoW) is one of the simplest and most widely used text vectorization techniques in Natural Language Processing (NLP). It converts text into numerical vectors by creating a vocabulary of unique words and counting the frequency of each word in a document. 

---

## How Bag of Words Works

1. Collect all text documents.
2. Perform text preprocessing:
   - Convert text to lowercase
   - Remove special characters
   - Remove stopwords
   - Apply stemming or lemmatization
3. Create a vocabulary of all unique words.
4. Count the frequency of each word in every document.
5. Generate a numerical feature matrix for Machine Learning models.

---

## Example

### Documents

```text
D1: I love Machine Learning
D2: Machine Learning is amazing
D3: I love Python
```

### Vocabulary

```text
[amazing, learning, love, machine, python]
```

### Bag of Words Matrix

| Document | amazing | learning | love | machine | python |
|----------|---------:|---------:|-----:|--------:|-------:|
| D1 | 0 | 1 | 1 | 1 | 0 |
| D2 | 1 | 1 | 0 | 1 | 0 |
| D3 | 0 | 0 | 1 | 0 | 1 |

---

## Advantages

-  Simple and easy to understand.
-  Easy to implement using `CountVectorizer`.
-  Fast and computationally efficient.
-  Works well for text classification tasks like Spam Detection.
-  Converts text into numerical vectors that Machine Learning models can process.
-  Suitable for small and medium-sized datasets.

---

## Disadvantages

-  Ignores the order of words.
-  Cannot understand the meaning or context of words.
-  Produces sparse and high-dimensional feature matrices.
-  Treats synonyms as different words.
-  Cannot capture semantic relationships between words.
-  Vocabulary size increases with larger datasets.

---

## Applications

- Spam Email Detection
- Sentiment Analysis
- Fake News Detection
- Text Classification
- Document Categorization
- Information Retrieval
- NLP Preprocessing

---

## Conclusion

Bag of Words is a simple and effective text vectorization technique that converts text into numerical vectors based on word frequency. Although it ignores word order and semantic meaning, it serves as a strong baseline for many Natural Language Processing and Machine Learning applications.

````python
import re                             ## regular expression (remove the unwanted special characters and numbers)
````

````python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
````

````python
from sklearn.feature_extraction.text import CountVectorizer
````

````python
messages = ['The boy is good boy ','the girl is good girl ', 'the boy and girl are playing cricket']
````

````python
corpus =[]
ps = PorterStemmer()
````

````python
for sentence in messages:
    ## 1. step is remove unneccessary special characters 
    review = re.sub('[^a-zA-Z]', " ",sentence)              ## re--> regular , .sub --> substitute

    ## 2. step is lowerCase
    review = review.lower()

    ## 3. step is splitting the sentence into words (tokenize)
    review = review.split() 

    ## 4. Remove stopwords + Stemming
    review = [ps.stem(word)
              for word in review
              if word not in stopwords.words('english')]
    ## 5. join the words again
    review = " ".join(review)

    corpus.append(review)

    
````

````python
print(corpus)
````

### CREATING THE BAG OF WORDS ------

````python
from sklearn.feature_extraction.text import CountVectorizer
````

````python
cv = CountVectorizer()
x = cv.fit_transform(corpus).toarray()                     ## fit --> unique vocabulary , transform --> numbering the words 
print(cv.get_feature_names_out())                         ## get_features_names_out --> columns ke naam 
print(x)
````
